```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```


# Load data

```python
df = pd.read_csv('../data/raw/survey_data_2026.csv')
print(f"Total responses: {len(df)}")
```

# %%
# Display first few rows

```python
df.head()
```


# %%
# PHQ-9 score distribution

```python
phq_cols = ['phq9_1', 'phq9_2', 'phq9_3', 'phq9_4', 'phq9_5', 
            'phq9_6', 'phq9_7', 'phq9_8', 'phq9_9']
```

```python

df['phq9_total'] = df[phq_cols].sum(axis=1)
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df['phq9_total'], bins=20, edgecolor='black')
plt.xlabel('PHQ-9 Total Score')
plt.ylabel('Number of Students')
plt.title('Distribution of Depression Risk Scores')
plt.show()
```

# %%
# Risk categories

```python
df['risk_tier'] = pd.cut(df['phq9_total'], 
                         bins=[0, 4, 9, 14, 19, 27],
                         labels=['Minimal', 'Mild', 'Moderate', 'Moderately Severe', 'Severe'])

risk_counts = df['risk_tier'].value_counts()
print(risk_counts)
```
